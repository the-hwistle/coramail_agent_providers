```python
from langchain.tools import tool
from langchain_ollama import ChatOllama
from langchain.agents import create_agent
from langchain_core.messages import HumanMessage


@tool(parse_docstring=True)
def multiply(a: float, b: float) -> float:
    """두 수를 곱합니다.

    Args:
        a: 첫 번째 숫자
        b: 두 번째 숫자
    """
    return a * b


@tool(parse_docstring=True)
def add(a: float, b: float) -> float:
    """두 수를 더합니다.

    Args:
        a: 첫 번째 숫자
        b: 두 번째 숫자
    """
    return a + b


@tool(parse_docstring=True)
def divide(a: float, b: float) -> float:
    """첫 번째 수를 두 번째 수로 나눕니다.

    Args:
        a: 나누어지는 숫자
        b: 나누는 숫자
    """
    if b == 0:
        raise ValueError("0으로 나눌 수 없습니다.")

    return a / b


@tool(parse_docstring=True)
def subtract(a: float, b: float) -> float:
    """첫 번째 수에서 두 번째 수를 뺍니다.

    Args:
        a: 첫 번째 숫자
        b: 두 번째 숫자
    """
    return a - b


tools = [add, multiply, divide, subtract]


# Ollama 로컬 모델
llm = ChatOllama(
    model="qwen3.5:2b ",
    base_url="http://localhost:11434",
    temperature=0,
)


agent = create_agent(
    model=llm,
    tools=tools,
    system_prompt=(
        "당신은 계산 도우미입니다. "
        "산술 계산은 반드시 제공된 도구를 사용하십시오. "
        "여러 단계의 계산은 앞 단계의 결과를 다음 도구 호출에 전달하십시오."
    ),
)


result = agent.invoke(
    {
        "messages": [
            HumanMessage(
                content="도구를 사용하여 103과 439를 더한 후, 그 결과를 4로 나누세요."
            )
        ]
    }
)


for message in result["messages"]:
    message.pretty_print()
```

```python
# Cell 1. 설정

import os
import json
import uuid
import requests
from dataclasses import dataclass
from typing import Any

from qdrant_client import QdrantClient
from qdrant_client.models import Distance, VectorParams, PointStruct
from langchain.tools import tool
from langchain.agents import create_agent
from langchain_core.messages import HumanMessage
from langchain_ollama import ChatOllama


@dataclass(frozen=True)
class Settings:
    ollama_base_url: str = os.getenv("OLLAMA_HOST", "http://localhost:11434").rstrip("/")
    embedding_model: str = os.getenv("CORAMAIL_EMBEDDING_MODEL", "bge-m3")
    agent_model: str = os.getenv("CORAMAIL_AGENT_MODEL", "qwen3.5:2b")

    qdrant_url: str = os.getenv("QDRANT_URL", "http://localhost:6333")
    qdrant_collection: str = os.getenv(
        "CORAMAIL_QDRANT_COLLECTION",
        "coramail_agent_docs_bge_m3",
    )
    vector_size: int = int(os.getenv("CORAMAIL_VECTOR_SIZE", "1024"))


settings = Settings()
settings
```

```python
# Cell 2. Ollama helper

def embed_text(text: str) -> list[float]:
    response = requests.post(
        f"{settings.ollama_base_url}/api/embed",
        json={
            "model": settings.embedding_model,
            "input": text,
        },
        timeout=60,
    )

    if response.status_code == 404:
        response = requests.post(
            f"{settings.ollama_base_url}/api/embeddings",
            json={
                "model": settings.embedding_model,
                "prompt": text,
            },
            timeout=60,
        )

    response.raise_for_status()
    data = response.json()

    if "embeddings" in data:
        vector = data["embeddings"][0]
    else:
        vector = data.get("embedding")

    if not vector:
        raise RuntimeError(f"embedding 결과가 비어 있습니다: {data}")

    if len(vector) != settings.vector_size:
        raise ValueError(
            f"임베딩 차원 불일치: expected={settings.vector_size}, actual={len(vector)}"
        )

    return vector


def ollama_chat(prompt: str, model: str | None = None) -> str:
    response = requests.post(
        f"{settings.ollama_base_url}/api/chat",
        json={
            "model": model or settings.agent_model,
            "messages": [{"role": "user", "content": prompt}],
            "stream": False,
            "options": {
                "temperature": 0,
                "num_predict": 400,
            },
        },
        timeout=120,
    )
    response.raise_for_status()
    return response.json()["message"]["content"]


v = embed_text("테스트 문장입니다.")
print("embedding dim:", len(v))
```

```python
# Cell 3. Qdrant 초기화

client = QdrantClient(
    url=settings.qdrant_url,
    timeout=30,
)

try:
    existing_collections = [c.name for c in client.get_collections().collections]
except Exception as exc:
    raise RuntimeError(
        f"Qdrant 서버에 연결할 수 없습니다: {settings.qdrant_url}\n"
        "서버 모드로 실행하려면 Qdrant가 먼저 떠 있어야 합니다.\n"
        "예:\n"
        "  docker run -p 6333:6333 -p 6334:6334 qdrant/qdrant\n"
        "또는 QDRANT_URL 환경변수를 실제 서버 주소로 설정하세요."
    ) from exc

if settings.qdrant_collection not in existing_collections:
    client.create_collection(
        collection_name=settings.qdrant_collection,
        vectors_config=VectorParams(
            size=settings.vector_size,
            distance=Distance.COSINE,
        ),
    )
    print("created:", settings.qdrant_collection)
else:
    print("exists:", settings.qdrant_collection)

print("qdrant_url:", settings.qdrant_url)
```

```python
# Cell 4. 문서 chunking / 색인 함수

def split_text(text: str, chunk_size: int = 1200, overlap: int = 150) -> list[str]:
    text = " ".join((text or "").split())
    if not text:
        return []

    chunks = []
    start = 0

    while start < len(text):
        end = start + chunk_size
        chunks.append(text[start:end])

        if end >= len(text):
            break

        start = max(0, end - overlap)

    return chunks


def stable_point_id(source_id: str, chunk_index: int) -> str:
    return str(uuid.uuid5(uuid.NAMESPACE_URL, f"{source_id}:{chunk_index}"))


def index_documents(documents: list[dict[str, Any]]) -> dict[str, Any]:
    points = []

    for doc in documents:
        source_id = str(doc.get("source_id") or doc.get("filename") or uuid.uuid4())
        title = str(doc.get("title") or "")
        text = str(doc.get("text") or "")

        chunks = split_text(text)

        for idx, chunk in enumerate(chunks):
            vector = embed_text(chunk)

            payload = {
                "source_id": source_id,
                "title": title,
                "chunk_index": idx,
                "text": chunk,
            }

            for key, value in doc.items():
                if key != "text":
                    payload[key] = value

            points.append(
                PointStruct(
                    id=stable_point_id(source_id, idx),
                    vector=vector,
                    payload=payload,
                )
            )

    if points:
        client.upsert(
            collection_name=settings.qdrant_collection,
            points=points,
        )

    return {
        "documents": len(documents),
        "points": len(points),
        "collection": settings.qdrant_collection,
        "embedding_model": settings.embedding_model,
    }
```
